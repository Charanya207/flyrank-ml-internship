## 3. Leakage audit

*The same hunt from Week 3, on your final feature set.*

I checked the final feature set for leakage. Client identifiers, the target label, and target-related future signals are excluded from the model features. This reduces the risk of the model using information that would not be available for a fair prediction.

```python
import os
import pandas as pd

# Find the dataset wherever it exists in the current Colab session
matches = []

for root, dirs, files in os.walk("/content"):
    if "content_refresh_anonymized.csv" in files:
        matches.append(os.path.join(root, "content_refresh_anonymized.csv"))

print("Dataset files found:")
for path in matches:
    print(path)

if len(matches) == 0:
    print("Dataset not found in this Colab session.")
else:
    df = pd.read_csv(matches[0])

    leakage_cols = [
        "content_id",
        "client_id",
        "trend_direction",
        "trend_pct",
        "is_declining_label"
    ]

    feature_cols = [c for c in df.columns if c not in leakage_cols]

    found_leakage = [col for col in leakage_cols if col in feature_cols]

    print("\nChecking for leakage columns...")
    print("Leakage columns found in features:", found_leakage)
    print("Leakage check passed:", len(found_leakage) == 0)
```
